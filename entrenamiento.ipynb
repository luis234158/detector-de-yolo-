## 2. Descargar el dataset desde Roboflow

Reemplaza `TU_API_KEY` por tu clave (Roboflow → Settings → API Keys). **No subas la clave a GitHub.**

```python
from roboflow import Roboflow

rf = Roboflow(api_key="TU_API_KEY")
project = rf.workspace("est-luism-ruiz-unimilitar-edu-co").project("detector-de-monedas-ue9vk")
dataset = project.version(2).download("yolov11")
```

## 3. Entrenar

- `yolo11s`: modelo pequeño, buen balance entre velocidad y precisión.
- `patience=30`: se detiene si no mejora en 30 épocas.

```python
from ultralytics import YOLO

model = YOLO("yolo11s.pt")
model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    patience=30,
)
```

## 4. Evaluar

```python
metrics = model.val()
print(f"mAP50:    {metrics.box.map50:.3f}")
print(f"mAP50-95: {metrics.box.map:.3f}")
```
